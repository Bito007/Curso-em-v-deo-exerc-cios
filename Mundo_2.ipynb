```python
# Exercicio 52 -------------------------------------------------------------
num = int(input('Numero: '))

divisoes = []
for i in range(1, num + 1):
  divisoes.append(i)

tentativas = 0
for n in range(1, num + 1):
  if num % n == 0:
    tentativas += 1
    divisoes[divisoes.index(n)] = f'\033[34m{n}\033[31m'

print(f'O numero {num} é divisivel {tentativas} vezes')
if tentativas >= 3:
  print('Não é primo')
else:
  print('Primo')

for i in divisoes:
  print(i, end=' ')
```

```python
# Exercicio 68 ----------------------------------------------------------------
import random

ganhas = 0
status = True
while status:
  computador_num = random.randint(0, 10)
  computador_modo = None

  numero = int(input('Numero: '))
  modo = str(input('Par ou impar?[P/I]: '))

  if modo == "P":
    computador_modo = "I"
  else:
    computador_modo = "P"

  soma = computador_num + numero
  resto = soma % 2

  if resto == 0 and modo == "P":
    print(f'Bot: {computador_num} & {computador_modo}')
    print('Deu Par!')
    print('Ganhou!')
    status = True
    ganhas += 1
  elif resto == 1 and modo == "I":
    print(f'Bot: {computador_num} & {computador_modo}')
    print('Deu Impar!')
    print('Ganhou!')
    status = True
    ganhas += 1
  else:
    print(f'Bot: {computador_num} & {computador_modo}')
    if resto == 1:
      print('Deu Impar!')
    elif resto == 0:
      print('Deu Par!')
    print('Perdeu!')
    status = False

print(f'Partidas ganhas: {ganhas}')


```

```python
# Exercicio 71 ----------------------------------------------------------------
valor = int(input('Valor: '))
total = valor
cedulas_50 = 0
cedulas_20 = 0
cedulas_10 = 0
cedulas_1 = 0
while True:
  if total >= 50:
    total -= 50
    cedulas_50 += 1
  elif total >= 20:
    total -= 20
    cedulas_20 += 1
  elif total >= 10:
    total -= 10
    cedulas_10 += 1
  elif total >= 1:
    total -= 1
    cedulas_1 += 1
  else:
    break

if cedulas_50 > 0:
  print(f'Total de {cedulas_50} de R$50')
if cedulas_20 > 0:
  print(f'Total de {cedulas_20} de R$20')
if cedulas_10 > 0:
  print(f'Total de {cedulas_10} de R$10')
if cedulas_1 > 0:
  print(f'Total de {cedulas_1} de R$1')

```
